# Blinking Leds

Simple program that turns the pynq's leds on and off 1000 times and then finally turns them off.

```python
from pynq.overlays.base import BaseOverlay

import time

base = BaseOverlay("base.bit")

try:
    for _ in range(1000):
        for led in base.leds:
            led.on()
        time.sleep(0.5)

        for led in base.leds:
            led.off()
        time.sleep(0.5)
finally:
    for led in base.leds:
        led.off()

print("Finished blinking")
```
